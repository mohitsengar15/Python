Question 1: What is Tableau? Explain its importance in Business Intelligence and how it helps in data-driven decision-making.

Answer:
Tableau is a powerful data visualization and Business Intelligence (BI) tool used to analyze, visualize, and share data in the form of interactive dashboards and reports. It allows users to connect to multiple data sources, transform raw data, and create meaningful insights without requiring deep programming knowledge.

In Business Intelligence, Tableau plays a crucial role by converting complex datasets into easy-to-understand visual formats such as charts, graphs, and dashboards. This helps organizations quickly identify trends, patterns, and outliers.

It supports data-driven decision-making by enabling:

Real-time data analysis
Interactive dashboards for deeper exploration
Better communication of insights to stakeholders
Faster and more accurate business decisions
Question 2: Explain the role of the following Tableau components:

a) Data Pane
b) Worksheet
c) Dashboard
d) Story

Answer:

a) Data Pane:
The Data Pane is where all the fields (dimensions and measures) of the dataset are displayed. It helps users drag and drop fields into the workspace to create visualizations.

b) Worksheet:
A Worksheet is the main working area where individual charts or visualizations are created using data fields.

c) Dashboard:
A Dashboard is a collection of multiple worksheets combined into a single interactive view. It allows users to compare different visualizations together.

d) Story:
A Story is a sequence of dashboards or worksheets arranged to present a narrative or insights step-by-step.

Question 3: What is the difference between Dimensions and Measures in Tableau? Provide examples of each.

Answer:

Dimensions:
Dimensions are qualitative data fields used to categorize and segment data. They are usually non-numeric.
Examples: Country, Product Name, Category, Date

Measures:
Measures are quantitative data fields used for calculations and aggregation. They are numeric values.
Examples: Sales, Profit, Quantity, Revenue

Difference:

Dimensions describe data
Measures quantify data
Question 4: Define and explain the purpose of Filters, Parameters, and Sets in Tableau.

Answer:

Filters:
Filters are used to limit the data displayed in a visualization based on specific conditions.
Example: Showing sales for only one country.

Parameters:
Parameters are dynamic inputs that allow users to change values in calculations or visualizations.
Example: Switching between Profit and Sales.

Sets:
Sets are custom subsets of data based on conditions.
Example: Top 10 customers by sales.

Question 6: Create a dual-axis chart (Monthly Sales & Profit for 2014)

Answer:
Steps:

Drag Date (Month) to Columns
Drag Sales to Rows (Bar chart)
Drag Profit to Rows → Right-click → Dual Axis
Synchronize axes
Change Profit to Line chart
Filter Year = 2014
Add title and format

Question 1: Difference between descriptive and inferential statistics

Answer:

Descriptive Statistics: Summarizes and describes data using measures like mean, median, and graphs.
Example: Average marks of a class.
Inferential Statistics: Uses sample data to make predictions or conclusions about a population.
Example: Predicting election results using surveys.
Question 2: What is sampling? Random vs stratified sampling

Answer:
Sampling is the process of selecting a subset from a population to study.

Random Sampling: Every individual has equal chance of selection.
Stratified Sampling: Population is divided into groups (strata), and samples are taken from each group.

Question 5: Python – Mean, Median, Mode

Answer:

```python
import statistics

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

mean = statistics.mean(numbers)
median = statistics.median(numbers)
mode = statistics.mode(numbers)

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
```

Question 6: Covariance and Correlation

Answer:

```python
import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

covariance = np.cov(list_x, list_y)[0][1]
correlation = np.corrcoef(list_x, list_y)[0][1]

print("Covariance:", covariance)
print("Correlation:", correlation)
```

Question 8: Advertising vs Sales (Correlation)

Answer:

Explanation:

Covariance: Shows direction of relationship
Correlation: Shows strength (-1 to +1)

```python
import numpy as np

advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

correlation = np.corrcoef(advertising_spend, daily_sales)[0][1]

print("Correlation:", correlation)
```

Question 9: Survey Data Analysis

Answer:

Explanation:
Use:

Mean → average satisfaction
Standard deviation → variation
Histogram → distribution

```python
import matplotlib.pyplot as plt

survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

plt.hist(survey_scores, bins=5)
plt.title("Survey Scores Distribution")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.show()
```
